```python
import pandas as pd
import numpy as np
from datetime import datetime
from scipy import stats
import seaborn as sns 
import matplotlib.pyplot as plt 
```

Commande qui charge le Dataset

```python
df=pd.read_csv("Dataset/Listing_AirBnB_Paris.csv")
```

Nettoyages de données

```python
df_clean = df.copy()
df_clean.to_csv("Dataset/Listing_AirBnB_Paris_clean.csv", index=False)
```

```python
valuesrows_with_nan = df_clean[df_clean.isna().any(axis=1)] #Visulaise les données avec pour entrées NaN
df_clean.drop(columns=['neighbourhood_group'], inplace=True, errors='ignore')#On supprime la colonne car elle ne possede que des NaN et est don obselète
print(valuesrows_with_nan)
print(len(valuesrows_with_nan))#Affiche le nombre total de lignes avec NaN
```

```python
print(df_clean.columns) #on verifie si la colonne est toujours présente
```

```python
#Je sélectionne certaines colonnes pour pouvoir facilement observer les Nan car la Dataset contient beaucoup de dataframe
valuesrows_with_nan = df_clean[df_clean.isna().any(axis=1)]
valuesrows_with_nan[['id', 'price', 'room_type', 'last_review']]

# Remplacement des NaN sur df_clean
df_clean['name'] = df_clean['name'].fillna('Inconnu')
df_clean['host_name'] = df_clean['host_name'].fillna('Inconnu')

# On remplace les avis par 0 si NaN
df_clean['number_of_reviews'] = df_clean['number_of_reviews'].fillna(0)
df_clean['number_of_reviews_ltm'] = df_clean['number_of_reviews_ltm'].fillna(0)
df_clean['reviews_per_month'] = df_clean['reviews_per_month'].fillna(0)

# On remplace la licence par "Non déclaré" si NaN
df_clean['license'] = df_clean['license'].fillna('Non déclaré')

# Suppression des lignes sans prix
df_clean = df_clean.dropna(subset=['price'])

```

```python
DataFrameduplicate_rows = df[df.duplicated()] #Identifie les doublons 
print(DataFrameduplicate_rows)

#L'index est vide car le Dataset ne contient pas de doublons
```

```python
# Traitement des valeurs aberrantes (on as utilisé ici la methode du calcul du score Z)
numeric_df = df_clean.select_dtypes(include=[np.number])# Sélectionne uniquement les colonnes numériques
z_scores = np.abs(stats.zscore(numeric_df))# Calcul du Z-score sur tout le dataset numérique
threshold = 3# Définir le seuil (souvent 3) 
outliers = (z_scores > threshold)# Détecter les outliers
print(outliers.sum())# Compter combien il y en a par colonne
```

Analyse Et Visualisation

```python
#Graphique boxplot représentant la fluctuation de prix en fonction du type de logement 
room_types = df_clean['room_type'].unique()
D = [df_clean[df_clean['room_type'] == room]['price'] for room in room_types] #une liste de séries de prix (une par type de logement)


# plot
fig, ax = plt.subplots(figsize=(8,6))
VP = ax.boxplot(D, positions = np.arange(len(D)) + 1 , widths=0.6, patch_artist=True,
                showmeans=False, showfliers=False,
                medianprops={"color": "white", "linewidth": 1.5},
                boxprops={"facecolor": "C0", "edgecolor": "white","linewidth": 1},
                whiskerprops={"color": "C0", "linewidth": 1.2},
                capprops={"color": "C0", "linewidth": 1.2})


ax.set_xticks(np.arange(len(D)) + 1)
ax.set_xticklabels(room_types, rotation=10)#affiche les noms des types de logement sous chaque boîte.
ax.set_ylabel("Prix (€)")
ax.set_title("Fluctuation du prix selon le type de logement - Airbnb Paris")

plt.show()
```

Le boxplot ci-dessus représente la distribution du prix selon le type de logement.
On observe que les logements entiers et les chambres d’hôtel affichent les prix les plus élevés,
tandis que les chambres partagées ont des tarifs plus bas.


Entire home/apt: fourchette compris entre 130-150 eu;
Private room: fourchette compris entre 80-90 eu;
Shared room: fourchette compris entre 30-40 eu;
Hotel room: fourchette compris entre 210-230 eu;

```python
#Graphique boxplot représentant la fluctuation de prix en fonction du quartier(neighbourhood) /uniquement les 10 les plus fréquent

top_neigh = df_clean['neighbourhood'].value_counts().head(10).index # Sélectionner les 10 quartiers les plus fréquents
D = [df_clean[df_clean['neighbourhood'] == rue]['price'] for rue in top_neigh ] #une liste de séries de prix (une par quartier )


# plot
fig, ax = plt.subplots(figsize=(14,8))
VP = ax.boxplot(D, positions = np.arange(len(D)) + 1 , widths=0.6, patch_artist=True,
                showmeans=False, showfliers=False,
                medianprops={"color": "white", "linewidth": 1.5},
                boxprops={"facecolor": "C0", "edgecolor": "white","linewidth": 1},
                whiskerprops={"color": "C0", "linewidth": 1.2},
                capprops={"color": "C0", "linewidth": 1.2})


ax.set_xticks(np.arange(len(D)) + 1)
ax.set_xticklabels(top_neigh, rotation=50)#affiche les noms des quarties sous chaque boîte.
ax.set_ylabel("Prix (€)")
ax.set_title("Fluctuation du prix selon le top 10 des quartier retrouver le plus fréquemment - Airbnb Paris")

plt.show()
```

On remarque que des villes comme Passy, Opéra,Temple,Vaugirard et Batignolles-Monceau ont des prix relativement hauts par rapports aux autres , cela atteste de leur attractivités au près des voyageurs 

```python

#Graphique heatmap représenatnt la répartition des logment selon disponibilités et prix
df_clean['price_cat'] = pd.cut(df_clean['price'], bins=[0,100,200,300,500,1000], labels=['<100','100-200','200-300','300-500','>500']) #détermines les classe de prix
df_clean['dispo_cat'] = pd.cut(df_clean['availability_365'], bins=[0,100,200,300,365], labels=['0-100','100-200','200-300','300-365'])#détermine les periodes de disponibilités s'étalant sur l'année

plt.figure(figsize=(8,6)) #taille du tableau/graphique
sns.heatmap(pd.crosstab(df_clean['price_cat'], df_clean['dispo_cat']), cmap="YlGnBu", annot=True, fmt="d") #plus le nombre de disponibilité est grand , plus la couleur est foncée
plt.title("Répartition du nombre de logements selon le prix et la disponibilité - Airbnb Paris")
plt.xlabel("Disponibilité (jours)")
plt.ylabel("Catégorie de prix (€)")
plt.show()

```

On remarque ici que les logement situer dans la fourchettes de prix 100-200 euros sont généralement beaucoup plus disponible et que les logement au dessus de 500 euro sont rarement disponible , attestant de leur popularité au près des clients

```python

#Graphique heatmap représentant la corrélation entre les variables numériques principales
plt.figure(figsize=(8,6)) #taille du graphique
corr = df_clean[['price', 'availability_365', 'reviews_per_month', 'minimum_nights', 'number_of_reviews']].corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='YlGnBu', fmt=".2f") #Les cases du tableau représentent un coefficient de corrélation compris entre -1 (correlation négative)et 1(correlation parfaite) 0 (aucune relation entre 2 variables)

plt.title("Corrélation entre les principales variables numériques - Airbnb Paris")
plt.show()
```
